```python
dbutils.widgets.text("catalogo", "workspace", "Catálogo")
dbutils.widgets.text("esquema", "demo_incremental", "Esquema")

catalogo = dbutils.widgets.get("catalogo")
esquema  = dbutils.widgets.get("esquema")

print("1️⃣ CAPA BRONCE (evidencia cruda inmutable):")
display(spark.sql(f"""
    SELECT id
        , nombre
        , ciudad
        , fecha_cambio
        , entrega
        , archivo
        , createat
    FROM {catalogo}.{esquema}.bronce_clientes
    ORDER BY entrega ASC, id ASC
"""))

print("2️⃣ CAPA PLATA - SCD TIPO 1 (fotografía actual sin historial):")
display(spark.sql(f"""
    SELECT *
    FROM {catalogo}.{esquema}.clientes_scd1
    ORDER BY id ASC
"""))

print("3️⃣ CAPA PLATA - SCD TIPO 2 (trazabilidad histórica completa):")
display(spark.sql(f"""
    SELECT *
    FROM {catalogo}.{esquema}.clientes_scd2
    ORDER BY id ASC, fecha_inicio ASC
"""))

print("4️⃣ CAPA ORO (métrica agregada vigente por ciudad):")
display(spark.sql(f"""
    SELECT ciudad
        , clientes_vigentes
        , fecha_calculo
    FROM {catalogo}.{esquema}.oro_clientes_por_ciudad
    ORDER BY clientes_vigentes DESC, ciudad ASC
"""))
```
